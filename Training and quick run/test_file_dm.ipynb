```python
import tarfile
import numpy as np
import matplotlib.pyplot as plt
```

```python
dm_list = []

tar_paths = ["batch_18.tar"]

tars = [tarfile.open(p) for p in tar_paths]

for tar_path in tar_paths:
    with tarfile.open(tar_path, "r") as tar:
        for member in tar:
            if member.isfile() and member.name.endswith(".npz"):
                with tar.extractfile(member) as f:
                    data = np.load(f)
                    dm = float(data['params'][0])

                    #print(member.name)
                    #print(data.files)
                    #print(f"DM: {dm}")
                    dm_list.append(dm)

plt.hist(dm_list, bins=14, edgecolor='black')
plt.xlabel("Dispersion Measure (pc cm$^{-3}$)")
plt.ylabel("Frequency")
```
